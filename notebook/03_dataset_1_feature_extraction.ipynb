```python

def inspect_schema(value, depth=0):

    indent = "  " * depth

    if isinstance(value, dict):

        for key, item in value.items():

            print(
                f"{indent}{key}: "
                f"{type(item).__name__}",
                end=""
            )

            if isinstance(item, (list, dict)):
                print(f" | Length: {len(item)}")

                if isinstance(item, list) and item:
                    print(
                        f"{indent}  Element type: "
                        f"{type(item[0]).__name__}"
                    )

                    if isinstance(item[0], dict):
                        print(
                            f"{indent}  Element keys: "
                            f"{list(item[0].keys())}"
                        )

                elif isinstance(item, dict):
                    print(
                        f"{indent}  Keys: "
                        f"{list(item.keys())[:20]}"
                    )

            else:
                print(f" | Sample: {str(item)[:100]}")


for col in complex_columns:

    print("\n" + "=" * 75)
    print("COLUMN:", col.upper())
    print("=" * 75)

    sample = next(
        (
            value for value in df_structured[col]
            if isinstance(value, (dict, list)) and len(value) > 0
        ),
        None
    )

    if sample is not None:
        inspect_schema(sample)
    else:
        print("No non-empty sample")
```

```python

# ============================================
# STEP 10: STRING AND GENERAL FEATURE EXTRACTION
# ============================================

# Create a separate feature DataFrame.
# Original structured dataset remains unchanged.

df_features = pd.DataFrame(index=df_structured.index)

# Preserve original target label.
df_features["label"] = df_structured["label"].copy()


# --------------------------------------------
# 1. Extract STRING features
# --------------------------------------------

def safe_dict(value):
    return value if isinstance(value, dict) else {}


string_data = df_structured["strings"].apply(safe_dict)

# Basic string statistics
string_basic_features = [
    "numstrings",
    "avlength",
    "printables",
    "entropy"
]

for feature in string_basic_features:

    df_features[f"strings_{feature}"] = (
        string_data.apply(lambda x: x.get(feature))
    )


# Expand string_counts dictionary
string_count_data = string_data.apply(
    lambda x: safe_dict(x.get("string_counts"))
)

# Identify all available string-count keys
string_count_keys = sorted(
    set().union(
        *(set(item.keys()) for item in string_count_data)
    )
)

# Extract each key into a separate feature
for key in string_count_keys:

    df_features[f"strings_count_{key}"] = (
        string_count_data.apply(
            lambda x: x.get(key, 0)
        )
    )


# --------------------------------------------
# 2. Extract GENERAL features
# --------------------------------------------

general_data = df_structured["general"].apply(safe_dict)

general_basic_features = [
    "size",
    "entropy",
    "is_pe"
]

for feature in general_basic_features:

    df_features[f"general_{feature}"] = (
        general_data.apply(lambda x: x.get(feature))
    )


# --------------------------------------------
# 3. Extract first four bytes
# --------------------------------------------

def extract_start_byte(value, position):

    if isinstance(value, (list, tuple)) and len(value) > position:
        return value[position]

    return np.nan


for position in range(4):

    df_features[f"general_start_byte_{position}"] = (
        general_data.apply(
            lambda x: extract_start_byte(
                x.get("start_bytes"),
                position
            )
        )
    )


# --------------------------------------------
# 4. Check MZ signature
# --------------------------------------------

df_features["general_has_mz_signature"] = (
    general_data.apply(
        lambda x: (
            isinstance(x.get("start_bytes"), (list, tuple))
            and len(x["start_bytes"]) >= 2
            and x["start_bytes"][0:2] == [77, 90]
        )
    ).astype("int8")
)


# --------------------------------------------
# 5. Convert extracted features to numeric
# --------------------------------------------

feature_columns = [
    col for col in df_features.columns
    if col != "label"
]

df_features[feature_columns] = (
    df_features[feature_columns]
    .apply(pd.to_numeric, errors="coerce")
)


# --------------------------------------------
# 6. Generate extraction summary
# --------------------------------------------

print("STRING AND GENERAL FEATURE EXTRACTION COMPLETED")

print("Original dataset shape:", df_structured.shape)

print("Extracted feature dataset shape:", df_features.shape)

print("Number of extracted features:", len(feature_columns))

print("\nLabel distribution:")
print(df_features["label"].value_counts())

print("\nMissing values:")
print(
    df_features[feature_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

print("\nExtracted feature preview:")
display(df_features.head(5))
```

3. Code – Extract HEADER and SECTION features

```python

# STEP 11: HEADER AND SECTION FEATURE EXTRACTION

import pandas as pd
import numpy as np
from collections import Counter

# --------------------------------------------------
# 1. HELPER FUNCTION: SAFE VALUE EXTRACTION
# --------------------------------------------------

def safe_dict(value):
    return value if isinstance(value, dict) else {}


# --------------------------------------------------
# 2. HEADER FEATURE EXTRACTION
# --------------------------------------------------

def extract_header_features(header):

    header = safe_dict(header)

    features = {}

    # COFF header
    coff = safe_dict(header.get("coff"))

    for key, value in coff.items():

        if isinstance(value, list):
            features[f"header_coff_{key}_count"] = len(value)

            for item in value:
                features[f"header_coff_{key}_{item}"] = 1

        elif isinstance(value, (int, float, bool)):
            features[f"header_coff_{key}"] = value

        elif isinstance(value, str):
            features[f"header_coff_{key}"] = value

    # Optional header
    optional = safe_dict(header.get("optional"))

    for key, value in optional.items():

        if isinstance(value, (int, float, bool)):
            features[f"header_optional_{key}"] = value

        elif isinstance(value, str):
            features[f"header_optional_{key}"] = value

    # DOS header
    dos = safe_dict(header.get("dos"))

    for key, value in dos.items():

        if isinstance(value, (int, float, bool)):
            features[f"header_dos_{key}"] = value

    return features


# --------------------------------------------------
# 3. SECTION FEATURE EXTRACTION
# --------------------------------------------------

def extract_section_features(section):

    section = safe_dict(section)

    features = {}

    # Entry section
    entry = section.get("entry")

    if isinstance(entry, str):
        features["section_entry"] = entry

    # Individual PE sections
    sections = section.get("sections", [])

    if not isinstance(sections, list):
        sections = []

    features["section_count"] = len(sections)

    sizes = []
    virtual_sizes = []
    entropies = []
    size_ratios = []
    vsize_ratios = []

    property_counts = Counter()
    section_name_counts = Counter()

    for item in sections:

        if not isinstance(item, dict):
            continue

        name = item.get("name")

        if isinstance(name, str):
            section_name_counts[name] += 1

        # Numerical section properties
        for key, target in [
            ("size", sizes),
            ("vsize", virtual_sizes),
            ("entropy", entropies),
            ("size_ratio", size_ratios),
            ("vsize_ratio", vsize_ratios)
        ]:

            value = item.get(key)

            if isinstance(value, (int, float)):
                target.append(value)

        # Section permission and characteristic flags
        props = item.get("props", [])

        if isinstance(props, list):
            property_counts.update(
                str(prop) for prop in props
            )

    # Aggregate numerical statistics
    collections = {
        "size": sizes,
        "vsize": virtual_sizes,
        "entropy": entropies,
        "size_ratio": size_ratios,
        "vsize_ratio": vsize_ratios
    }

    for name, values in collections.items():

        features[f"section_{name}_sum"] = sum(values)

        features[f"section_{name}_mean"] = (
            np.mean(values) if values else 0
        )

        features[f"section_{name}_max"] = (
            max(values) if values else 0
        )

        features[f"section_{name}_min"] = (
            min(values) if values else 0
        )

        features[f"section_{name}_std"] = (
            np.std(values) if len(values) > 1 else 0
        )

    # Section property frequencies
    for prop, count in property_counts.items():

        features[f"section_prop_{prop}_count"] = count

    # Section name frequencies
    for name, count in section_name_counts.items():

        features[f"section_name_{name}_count"] = count

    # Overlay information
    overlay = safe_dict(section.get("overlay"))

    for key, value in overlay.items():

        if isinstance(value, (int, float, bool)):
            features[f"section_overlay_{key}"] = value

    return features


# --------------------------------------------------
# 4. APPLY EXTRACTION TO ALL RECORDS
# --------------------------------------------------

header_features = pd.DataFrame(
    df_structured["header"].apply(
        extract_header_features
    ).tolist(),
    index=df_structured.index
)

section_features = pd.DataFrame(
    df_structured["section"].apply(
        extract_section_features
    ).tolist(),
    index=df_structured.index
)


# --------------------------------------------------
# 5. COMBINE FEATURES
# --------------------------------------------------

df_header_section = pd.concat(
    [header_features, section_features],
    axis=1
)

print("HEADER FEATURES:", header_features.shape)
print("SECTION FEATURES:", section_features.shape)
print("COMBINED FEATURES:", df_header_section.shape)

display(df_header_section.head())
```

STEP 28 — Encode categorical features
Our strategy:
Feature	Unique values	Encoding
header_coff_machine	2	One-hot
header_optional_subsystem	7	One-hot
section_entry	High cardinality	Top 30 + OTHER, then one-hot

```python

# STEP 28: CATEGORICAL ENCODING

categorical_cols = df_reduced.select_dtypes(
    exclude=np.number
).columns.tolist()

df_encoding = df_reduced.copy()

# Store category mappings for reproducibility
categorical_mappings = {}

for col in categorical_cols:

    # Normalize categorical values
    df_encoding[col] = df_encoding[col].astype(str)

    unique_count = df_encoding[col].nunique()

    # Handle high-cardinality features
    if unique_count > 30:

        top_categories = (
            df_encoding[col]
            .value_counts()
            .head(30)
            .index.tolist()
        )

        categorical_mappings[col] = top_categories

        df_encoding[col] = df_encoding[col].where(
            df_encoding[col].isin(top_categories),
            "__OTHER__"
        )

    else:
        categorical_mappings[col] = (
            df_encoding[col].unique().tolist()
        )

# One-hot encoding
df_encoded = pd.get_dummies(
    df_encoding,
    columns=categorical_cols,
    dtype=np.uint8
)

print("CATEGORICAL ENCODING COMPLETED")
print("Original shape:", df_reduced.shape)
print("Encoded shape:", df_encoded.shape)
print("Remaining missing:", df_encoded.isna().sum().sum())

print("\nENCODED CATEGORICAL FEATURES:")
print(list(categorical_mappings.keys()))

print("\nFINAL DATA TYPES:")
print(df_encoded.dtypes.value_counts())
```
