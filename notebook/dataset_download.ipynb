```python
from pathlib import Path
from datetime import datetime
import re

# Dataset directories
TRAIN_DIR = Path(
    "/root/.cache/kagglehub/datasets/weiweip/ember2024/versions/1/Win64_train"
)

TEST_DIR = Path(
    "/root/.cache/kagglehub/datasets/weiweip/ember2024/versions/1/Win64_test"
)

# Extract the starting date from each filename
def get_start_date(file_path):
    match = re.search(r"(\d{4}-\d{2}-\d{2})_", file_path.name)

    if not match:
        raise ValueError(f"Date not found in filename: {file_path.name}")

    return datetime.strptime(match.group(1), "%Y-%m-%d")


# Find and sort JSONL files chronologically
train_files = sorted(
    TRAIN_DIR.glob("*.jsonl"),
    key=get_start_date
)

test_files = sorted(
    TEST_DIR.glob("*.jsonl"),
    key=get_start_date
)

# Select the latest two files from each directory
selected_train = train_files[-2:]
selected_test = test_files[-2:]

print("SELECTED TRAIN FILES:")
for file in selected_train:
    print(file.name)

print("\nSELECTED TEST FILES:")
for file in selected_test:
    print(file.name)
```

```python
from google.colab import files
import pandas as pd

# We have the selected lists of files from the previous step:
# selected_train: containing the 2 latest training file paths
# selected_test: containing the 2 latest testing file paths

all_selected_files = list(selected_train) + list(selected_test)

for file_path in all_selected_files:
    # Define the output CSV name based on the original JSONL filename
    csv_name = file_path.stem + '.csv'
    print(f"Converting {file_path.name} to {csv_name}...")

    # Load and drop null values matching your preprocessing step
    df = pd.read_json(file_path, lines=True)
    df.dropna(inplace=True)

    # Save to CSV
    df.to_csv(csv_name, index=False)
    print(f"Saved. Downloading {csv_name}...")

    # Trigger individual browser download
    files.download(csv_name)
```

```python
from google.colab import files
import pandas as pd

# We will convert all 4 selected files again without dropping any null values
all_selected_files = list(selected_train) + list(selected_test)

for file_path in all_selected_files:
    csv_name = file_path.stem + '_with_nulls.csv'
    print(f"Converting {file_path.name} to {csv_name} (keeping all null values)...")

    # Load without dropping any rows
    df = pd.read_json(file_path, lines=True)

    # Save directly to CSV with null values intact
    df.to_csv(csv_name, index=False)
    print(f"Saved. Downloading {csv_name}...")

    # Trigger download
    files.download(csv_name)
```
